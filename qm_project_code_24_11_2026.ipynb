```python
import re
import requests

def get_movie_script(url):
    """
    Fetches and cleans a movie script from a given IMSDb HTML URL.
    """
    try:
        # Fetch webpage content
        response = requests.get(url)
        response.raise_for_status() # Raise an error if the request failed
        html_content = response.text

        # Extract text inside <pre>...</pre> tags
        match = re.search(r'<pre>(.*?)</pre>', html_content, re.DOTALL | re.IGNORECASE)

        if match:
            raw_script = match.group(1)
            # Remove HTML tags (like <b>, <i>, etc.)
            clean_script = re.sub(r'<[^>]+>', '', raw_script)
            return clean_script
        else:
            print("Error: Could not find the script block (<pre> tag) on this page.")
            return None

    except Exception as e:
        print(f"An error occurred while fetching the URL: {e}")
        return None

# Example usage:
aladdin_url = "https://imsdb.com/scripts/Aladdin.html"
aladdin_script = get_movie_script(aladdin_url)

# Preview the first 500 characters of the fetched script
if aladdin_script:
    print(aladdin_script[:500])
```

```python
import pandas as pd
import re
import requests

def download_sheet_data(sheet_url):
    """
    Converts a standard Google Sheets sharing link into a direct CSV export link
    and loads it into a pandas DataFrame.
    """
    # Convert edit link to export link
    csv_url = sheet_url.replace('/edit?gid=', '/export?format=csv&gid=')
    if '/export?format=csv' not in csv_url:
        csv_url = csv_url.split('/edit')[0] + '/export?format=csv'

    print(f"Downloading sheet data from: {csv_url}")
    df = pd.read_csv(csv_url)
    return df

# Test loading the spreadsheet
sheet_url = "https://docs.google.com/spreadsheets/d/1v6rLpeLWQa7HnlccEloSZSxrxOso9cYuPlFkHbg1yIY/edit?gid=0#gid=0"
try:
    scripts_df = download_sheet_data(sheet_url)
    display(scripts_df.head())
except Exception as e:
    print(f"Could not load Google Sheet directly (it might be private): {e}")
```

```python
def standardize_script(raw_script_text):
    """
    Analyzes the script structure and converts it to Aladdin's format:
    CHARACTER NAME:   [Stage Direction] Dialogue...
    """
    # Normalize Windows line endings (\r\n) to Unix style (\n) first
    normalized_text = raw_script_text.replace('\r\n', '\n').replace('\r', '\n')
    lines = normalized_text.split('\n')

    # Basic heuristic to check if the script is 'Anastasia style' (indented character names)
    # Count lines that have significant leading whitespace followed by uppercase names
    indented_name_pattern = re.compile(r'^[ ]{15,25}([A-Z]{2,}(?:\s+[A-Z]{2,})*)(?:\s*\(.*\))?$')
    indented_count = sum(1 for line in lines[:1000] if indented_name_pattern.match(line))

    # Check if it already matches Aladdin style (Names on the left with colons)
    colons_count = sum(1 for line in lines[:1000] if re.match(r'^[A-Z]{2,}(\s+[A-Z]{2,})*\s*:', line))

    print(f"Heuristic counts -> Indented Names: {indented_count}, Left Names with Colons: {colons_count}")

    if colons_count > indented_count and colons_count > 10:
        print("Detected Format: Standard (Aladdin-like) - No heavy conversion needed.")
        return normalized_text

    print("Detected Format: Indented/Centered (Anastasia-like) - Converting...")

    standardized_lines = []
    current_character = None

    for line in lines:
        # Check for character name pattern (indented with spaces, uppercase)
        name_match = indented_name_pattern.match(line)
        if name_match:
            current_character = name_match.group(1).strip()
            continue

        # If we have an active character, parse their dialogue
        if current_character:
            # If line is blank, it often ends dialogue or separates paragraphs
            if not line.strip():
                current_character = None
                standardized_lines.append("")
                continue

            # Clean dialogue lines of excess indentation
            dialogue_line = line.strip()

            # Check for embedded stage directions inside parentheses
            if dialogue_line.startswith("(") and dialogue_line.endswith(")"):
                standardized_lines.append(f"({dialogue_line[1:-1]})")
            else:
                standardized_lines.append(f"{current_character}:\t{dialogue_line}")
        else:
            # Line is action, stage direction, or scene heading
            stripped = line.strip()
            if stripped:
                if stripped.startswith("(") and stripped.endswith(")"):
                    standardized_lines.append(stripped)
                else:
                    # Format action/stage directions inside brackets / parentheses
                    standardized_lines.append(f"({stripped})")
            else:
                standardized_lines.append("")

    return "\n".join(standardized_lines)

# Test conversion on the Anastasia script loaded earlier
if 'anastasia_script' in globals() and anastasia_script:
    converted_anastasia = standardize_script(anastasia_script)
    print("\n--- PREVIEW OF CONVERTED ANASTASIA SCRIPT ---")
    print("\n".join(converted_anastasia.split('\n')[:50]))
```

```python
import re
from collections import Counter

def robust_standardize_script(raw_script_text):
    """
    A dynamic script parser that automatically detects character indentation levels
    and converts standard script formatting into the target unified layout:
    CHARACTER NAME:	Dialogue...
    (Action / Stage Directions)
    """
    if not raw_script_text:
        return ""

    # 1. Normalize line endings and tabs (convert tabs to 4 spaces for consistent counting)
    normalized_text = raw_script_text.replace('\r\n', '\n').replace('\r', '\n')
    normalized_text = normalized_text.expandtabs(4)
    lines = normalized_text.split('\n')

    # 2. Heuristically detect formatting type dynamically per script
    # Count leading spaces on uppercase-only lines to find the character margin
    indent_counts = []
    colon_lines_count = 0

    for line in lines[:2000]:  # Analyze first 2000 lines for layout signature
        stripped = line.strip()
        if not stripped:
            continue

        # Check for left-aligned names with colons (Aladdin style)
        if re.match(r'^[A-Z]{2,}(?:\s+[A-Z]{2,})*\s*:', stripped):
            colon_lines_count += 1
            continue

        # Match indented lines that look like CHARACTER NAMES (all uppercase, no ending punctuation)
        match = re.match(r'^([ ]+)([A-Z]{2,}(?:\s+[A-Z]{2,})*)(?:\s*\(.*\))?$', line)
        if match:
            indent_length = len(match.group(1))
            # Standard script character offsets are usually between 10 and 55 spaces
            if 10 <= indent_length <= 55:
                indent_counts.append(indent_length)

    # Determine layout strategy
    if colon_lines_count > 15:
        print("[Parser] Detected: Left-aligned with colons (No alignment adjustment needed).")
        return normalized_text

    if not indent_counts:
        # Fallback if no clean indentation patterns are found (e.g. Mary Poppins layout)
        print("[Parser] Detected: Unstructured/Unknown layout. Attempting lax margin detection...")
        # Fallback to general uppercase lines at any reasonable indent
        for line in lines[:1000]:
            match = re.match(r'^([ ]*)([A-Z]{2,}(?:\s+[A-Z]{2,})*)$', line)
            if match and len(match.group(1)) > 0:
                indent_counts.append(len(match.group(1)))

    # Find the most frequent indentation depth (the statistical Mode) for character names
    if indent_counts:
        common_character_indent = Counter(indent_counts).most_common(1)[0][0]
        print(f"[Parser] Detected: Indented Script layout. Character names are offset by ~{common_character_indent} spaces.")
    else:
        print("[Parser] Warning: Could not detect any consistent margin. Defaulting to standard 20 spaces.")
        common_character_indent = 20

    # 3. Parse and standardize the script
    standardized_lines = []
    current_character = None

    # Define margin tolerance (e.g. character name is exactly at the common indent +/- 2 spaces)
    min_indent = common_character_indent - 2
    max_indent = common_character_indent + 2

    for line in lines:
        stripped_line = line.strip()
        if not stripped_line:
            current_character = None
            standardized_lines.append("")
            continue

        # Check if this line matches the dynamically detected character name indent level
        leading_spaces = len(line) - len(line.lstrip(' '))

        # A line is a character name if it sits at the right margin, is uppercase, and doesn't contain script actions
        is_char_name = (
            min_indent <= leading_spaces <= max_indent and
            re.match(r'^[A-Z]{2,}(?:\s+[A-Z]{2,})*(?:\s*\(.*\))?$', stripped_line)
        )

        if is_char_name:
            # Extract name, ignoring parentheticals next to the name
            name_clean = re.sub(r'\(.*\)', '', stripped_line).strip()
            current_character = name_clean
            continue

        if current_character:
            # Process dialogue for the active character
            # Parentheticals on their own line under a character name
            if stripped_line.startswith("(") and stripped_line.endswith(")"):
                standardized_lines.append(f"({stripped_line[1:-1]})")
            else:
                standardized_lines.append(f"{current_character}:\t{stripped_line}")
        else:
            # Treat as action description or scene heading
            if stripped_line.startswith("(") and stripped_line.endswith(")"):
                standardized_lines.append(stripped_line)
            else:
                standardized_lines.append(f"({stripped_line})")

    return "\n".join(standardized_lines)

# Overwrite original function to apply our smart parser globally
standardize_script = robust_standardize_script
```

```python
def print_standardized_script_by_name(film_name, df):
    """
    Looks up a film in the DataFrame, fetches its script, standardizes the format,
    and prints the first 50 lines of the result.
    """
    # Find the correct column that contains 'Link' or 'URL' in its name
    link_col = [col for col in df.columns if 'link' in col.lower() or 'url' in col.lower()]
    name_col = [col for col in df.columns if 'name' in col.lower() or 'title' in col.lower()]

    if not link_col or not name_col:
        print("Error: Could not identify 'Film Name' or 'Link' columns in the spreadsheet.")
        print(f"Available columns are: {list(df.columns)}")
        return

    # Case-insensitive match for the film name
    match = df[df[name_col[0]].str.lower() == film_name.lower()]

    if match.empty:
        print(f"Film '{film_name}' not found in the spreadsheet.")
        # Suggest matches
        suggestions = df[df[name_col[0]].str.lower().str.contains(film_name.lower(), na=False)][name_col[0]].tolist()
        if suggestions:
            print(f"Did you mean one of these: {suggestions}?")
        return

    script_url = match.iloc[0][link_col[0]]
    actual_film_name = match.iloc[0][name_col[0]]

    print(f"Fetching script for '{actual_film_name}' from: {script_url}")
    raw_script = get_movie_script(script_url)

    if raw_script:
        standardized = standardize_script(raw_script)
        print(f"\n--- {actual_film_name.upper()} STANDARDIZED SCRIPT PREVIEW (First 50 lines) ---")
        print("\n".join(standardized.split('\n')[:50]))
        return standardized
    else:
        print("Failed to retrieve or clean the script.")
        return None

# Let's test it by passing one of the names from your spreadsheet!
# You can change 'Anastasia' to any movie present in your Google Sheet
try:
    converted_script = print_standardized_script_by_name("Anastasia", scripts_df)
except NameError:
    print("Please run the cell that downloads the sheet first to populate 'scripts_df'.")
```

```python
import os
import time
import zipfile
from google.colab import files

# Ensure scripts_df is loaded
if 'scripts_df' not in globals():
    print("scripts_df not found in memory. Attempting to load Google Sheet directly...")
    sheet_url = "https://docs.google.com/spreadsheets/d/1v6rLpeLWQa7HnlccEloSZSxrxOso9cYuPlFkHbg1yIY/edit?gid=0#gid=0"
    try:
        scripts_df = download_sheet_data(sheet_url)
    except NameError:
        # Redefine download_sheet_data if it's also missing from scope
        import pandas as pd
        def download_sheet_data(url):
            csv_url = url.replace('/edit?gid=', '/export?format=csv&gid=')
            if '/export?format=csv' not in csv_url:
                csv_url = csv_url.split('/edit')[0] + '/export?format=csv'
            return pd.read_csv(csv_url)
        scripts_df = download_sheet_data(sheet_url)

def batch_standardize_all_scripts(df, output_dir="standardized_scripts"):
    """
    Loops through all rows in the dataframe, downloads each script,
    standardizes it, and saves it as a text file.
    """
    # Create output directory
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # Identify columns
    link_col = [col for col in df.columns if 'link' in col.lower() or 'url' in col.lower()][0]
    name_col = [col for col in df.columns if 'name' in col.lower() or 'title' in col.lower()][0]

    successful_downloads = 0
    total_scripts = len(df)

    print(f"Starting batch processing for {total_scripts} scripts...\n")

    for idx, row in df.iterrows():
        film_name = row[name_col]
        url = row[link_col]

        # Clean filename for saving
        safe_filename = "".join([c for c in film_name if c.isalnum() or c in (' ', '_', '-')]).rstrip()
        file_path = os.path.join(output_dir, f"{safe_filename}_standardized.txt")

        print(f"[{idx+1}/{total_scripts}] Processing '{film_name}'...")

        try:
            # Fetch the raw script
            raw_script = get_movie_script(url)
            if not raw_script:
                print(f"  -> Skip: Failed to download script.")
                continue

            # Standardize using our robust parser
            standardized = standardize_script(raw_script)

            # Save to text file
            with open(file_path, "w", encoding="utf-8") as f:
                f.write(standardized)

            print(f"  -> Saved successfully to {file_path}")
            successful_downloads += 1

            # Polite pause to avoid overloading IMSDb servers
            time.sleep(1.0)

        except Exception as e:
            print(f"  -> Error processing '{film_name}': {e}")

    print(f"\nBatch processing complete! Successfully standardized {successful_downloads} out of {total_scripts} scripts.")

# Run the batch process
batch_standardize_all_scripts(scripts_df)
```
